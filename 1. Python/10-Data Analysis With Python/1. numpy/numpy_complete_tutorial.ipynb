### 1. Views vs. Copies (The Classic Memory Trap)
**Interview Question:** *What is the difference between a view and a copy in NumPy? How can you check if an array shares memory with another array?*

- **View:** A new array pointing to the *same* memory buffer. Slicing (`arr[1:4]`) creates views. Mutating a view modifies the original array!
- **Copy:** A completely separate memory allocation. Fancy indexing (`arr[[0, 2]]`) or boolean masking creates copies.
- **How to verify:** Inspect the `.base` attribute (returns `None` if it owns its memory) or `np.shares_memory(a, b)`.

```python
# Views vs Copies Demonstration
import numpy as np

original = np.array([10, 20, 30, 40, 50])

# Slicing creates a VIEW
view_arr = original[1:4]
print("Is view_arr a view of original?", view_arr.base is original)
print("Do they share memory?", np.shares_memory(original, view_arr))

# Modifying view modifies original!
view_arr[0] = 999
print("Original after modifying view:", original)

# Fancy Indexing creates a COPY
copy_arr = original[[0, 2, 4]]
print("Is copy_arr a view?", copy_arr.base is original)
copy_arr[0] = -1
print("Original after modifying copy:", original) # Unchanged!
```

### 2. Memory Layout & Strides (`arr.strides`)
**Interview Question:** *How does NumPy represent multidimensional arrays in 1D memory? What are strides?*

- A NumPy array is a contiguous 1D block of bytes in C.
- **Strides** is a tuple indicating how many **bytes** you must jump in memory to advance by 1 element along each dimension.
- For a 2D `float64` (8 bytes) array of shape `(3, 4)`:
  - Row jump = $4 \times 8 = 32$ bytes
  - Column jump = $1 \times 8 = 8$ bytes
  - Strides = `(32, 8)`

```python
# Strides and Memory Layout
matrix = np.arange(12, dtype=np.int64).reshape(3, 4) # int64 = 8 bytes per item

print("Matrix:", matrix)
print("Shape:", matrix.shape)
print("Item size in bytes:", matrix.itemsize)
print("Strides (row step bytes, col step bytes):", matrix.strides)

# Transposing changes strides WITHOUT copying data:
transposed = matrix.T
print("Transposed strides:", transposed.strides)
print("Shares memory with original?", np.shares_memory(matrix, transposed))
```

### 3. Broadcasting Rules Demystified
**Interview Question:** *Explain the exact rules of NumPy broadcasting. Why does `(3,) + (3, 1)` produce a `(3, 3)` matrix instead of an error?*

**Broadcasting Rules:**
1. If arrays have different numbers of dimensions, prepend `1`s to the shape of the smaller array.
2. Two dimensions are compatible if they are **equal**, or **one of them is 1**.
3. If neither condition is met, a `ValueError: operands could not be broadcast together` is thrown.

```python
# Broadcasting Demonstration
# (3,) vector vs (3, 1) column vector
row_vec = np.array([1, 2, 3])          # shape: (3,)
col_vec = np.array([[10], [20], [30]])  # shape: (3, 1)

# row_vec is treated as (1, 3)
# (1, 3) + (3, 1) broadcasts to (3, 3) grid:
result = row_vec + col_vec
print("row_vec shape:", row_vec.shape)
print("col_vec shape:", col_vec.shape)
print("Broadcasted result (3, 3):", result)

# Expanding dimensions using np.newaxis or None:
a = np.array([1, 2, 3])
print("Shape after np.newaxis:", a[:, np.newaxis].shape) # (3, 1)
```

### 4. Vectorized Multi-Condition Logic: `np.select` vs `np.where`
**Interview Question:** *How do you vectorize nested if-elif-else statements in NumPy without Python loops?*

- `np.where(cond, x, y)` works great for binary `if-else`.
- `np.select(condlist, choicelist, default)` is the idiomatic, ultra-fast solution for multiple `if-elif-else` branches.

```python
# Assigning letter grades based on scores without any loops:
scores = np.array([95, 82, 67, 45, 88, 73])

conditions = [
    scores >= 90,
    (scores >= 80) & (scores < 90),
    (scores >= 70) & (scores < 80),
    (scores >= 60) & (scores < 70)
]

choices = ['A', 'B', 'C', 'D']

grades = np.select(conditions, choices, default='F')
for s, g in zip(scores, grades):
    print(f"Score {s} -> Grade {g}")
```

### 5. High-Performance Top-K Selection: `np.argpartition`
**Interview Question:** *You have an array of 10 million elements. How do you find the top 5 largest elements in $O(N)$ time instead of $O(N \log N)$?*

- `np.sort` / `np.argsort` sorts the entire array: $O(N \log N)$.
- `np.argpartition(arr, -k)` partitions the array around the $k$-th element in $O(N)$ average time using Introselect!
- The largest $k$ elements are placed in the last $k$ positions (unsorted amongst themselves).

```python
# Top-K using np.argpartition vs np.argsort
data_arr = np.random.RandomState(42).randint(1, 1000, size=20)
k = 3

print("Data:", data_arr)

# Top k indices with O(N) partitioning:
top_k_idx = np.argpartition(data_arr, -k)[-k:]

# Sort just the top k elements:
top_k_idx_sorted = top_k_idx[np.argsort(-data_arr[top_k_idx])]

print(f"Top {k} indices:", top_k_idx_sorted)
print(f"Top {k} values:", data_arr[top_k_idx_sorted])
```

### 6. Vectorized Pairwise Euclidean Distance Matrix (No Loops!)
**Interview Question:** *Given $N$ points in $D$ dimensions represented as an $(N, D)$ array, compute the $(N, N)$ pairwise Euclidean distance matrix without using any Python for loops.*

Using broadcasting:
$$\text{dist}(i, j) = \sqrt{\sum_{d=1}^D (X_{i, d} - X_{j, d})^2}$$
Shape expansion: $(N, 1, D) - (1, N, D) \to (N, N, D)$.

```python
# Pairwise Euclidean Distance Matrix
points = np.array([
    [0, 0],
    [3, 4],
    [1, 1],
    [0, 5]
]) # Shape (4, 2): 4 points in 2D space

# Compute differences using (4, 1, 2) - (1, 4, 2) -> (4, 4, 2)
diff = points[:, np.newaxis, :] - points[np.newaxis, :, :]

# Square differences, sum along feature axis, take square root
dist_matrix = np.sqrt(np.sum(diff ** 2, axis=-1))

print("Pairwise Distance Matrix (4x4):")
print(np.round(dist_matrix, 2))
```

### 7. Linear Algebra: Systems of Equations & Matrix Inversion
**Interview Question:** *Why should you use `np.linalg.solve(A, b)` instead of computing `np.linalg.inv(A) @ b`?*

- **Numerical Stability:** Inverting a matrix explicitly is numerically unstable, prone to floating-point rounding errors, and computationally slower ($O(N^3)$ with large constant factor).
- `np.linalg.solve` uses LU decomposition / Gaussian elimination with partial pivoting, which is significantly more stable and accurate.

```python
# Solve 2x + 3y = 8 and 4x + 9y = 20
# A @ [x, y].T = b
A = np.array([[2, 3],
              [4, 9]], dtype=float)
b = np.array([8, 20], dtype=float)

# Recommended: np.linalg.solve
solution = np.linalg.solve(A, b)
print("Solution [x, y]:", solution)

# Verification:
print("A @ solution:", A @ solution) # Should equal [8, 20]
```

### 8. Moving Average via Convolution (`np.convolve`)
**Interview Question:** *How do you compute a rolling / moving average of window size $W$ over a 1D NumPy array in one line?*

A simple moving average is mathematically equivalent to the **discrete convolution** of the signal with a uniform rectangular window of weights $[1/W, 1/W, \dots, 1/W]$.

```python
# 1D Moving Average with np.convolve
signal = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90])
window_size = 3

weights = np.ones(window_size) / window_size
moving_avg = np.convolve(signal, weights, mode='valid')

print("Original Signal:", signal)
print(f"Moving Average (window={window_size}):", moving_avg)
```

### 9. Handling Missing Values (`np.nan`) & Imputation
**Interview Question:** *Why does `np.nan == np.nan` evaluate to `False`? How do you compute statistics ignoring `NaN`s and replace them?*

- In IEEE 754 floating-point standard, `NaN` is not equal to anything, including itself.
- Use `np.isnan()` to check for NaNs.
- Standard functions like `np.mean()` return `nan` if any NaN is present. Use `np.nanmean()`, `np.nanstd()`, `np.nanmedian()`.

```python
# NaN Handling & Imputation
arr_with_nans = np.array([12.0, np.nan, 18.0, np.nan, 25.0, 30.0])

print("arr == np.nan:", arr_with_nans == np.nan) # ALWAYS False!
print("np.isnan(arr):", np.isnan(arr_with_nans))

# Mean with and without ignoring NaNs:
print("Regular mean:", np.mean(arr_with_nans)) # returns nan
nan_mean = np.nanmean(arr_with_nans)
print("nanmean:", nan_mean)

# Vectorized NaN Imputation (replace NaNs with nanmean):
imputed = np.where(np.isnan(arr_with_nans), nan_mean, arr_with_nans)
print("Imputed Array:", imputed)
```
