# Categorical Data Plots

Now let's discuss using seaborn to plot categorical data! There are a few main plot types for this:

* boxplot
* violinplot
* stripplot
* swarmplot
* barplot
* countplot

Let's go through examples of each!

```python
df = sns.load_dataset("tips")
df
```

```python
sns.countplot(x = df['sex'],hue = df['smoker'])
```

```python
sns.barplot(x = df['sex'],y = df['total_bill'])
```

You can change the estimator object to your own function, that converts a vector to a scalar
Mean (default): np.mean
Median: np.median
Sum: np.sum
Count: len (to count the number of observations)
Standard Deviation: np.std
Minimum: np.min
Maximum: np.max

```python
import numpy as np 
sns.barplot(x = df['sex'],y = df['tip'],estimator=np.sum)
```

```python
sns.boxplot(x= 'tip',y = 'day',data = df,palette = 'rainbow')
```

## boxplot and violinplot

boxplots and violinplots are used to shown the distribution of categorical data. A box plot (or box-and-whisker plot) shows the distribution of quantitative data in a way that facilitates comparisons between variables or across levels of a categorical variable. The box shows the quartiles of the dataset while the whiskers extend to show the rest of the distribution, except for points that are determined to be “outliers” using a method that is a function of the inter-quartile range.

```python
sns.violinplot(x = 'day',y = 'total_bill',data = df)
```

## stripplot and swarmplot
The stripplot will draw a scatterplot where one variable is categorical. A strip plot can be drawn on its own, but it is also a good complement to a box or violin plot in cases where you want to show all observations along with some representation of the underlying distribution.

The swarmplot is similar to stripplot(), but the points are adjusted (only along the categorical axis) so that they don’t overlap. This gives a better representation of the distribution of values, although it does not scale as well to large numbers of observations (both in terms of the ability to show all the points and in terms of the computation needed to arrange them).

```python
sns.stripplot(x = 'day',y = 'total_bill',data = df)
```

```python
sns.swarmplot(x = 'day',y = 'total_bill',data = df)
```

```python
sns.violinplot(x= 'tip',y = 'day',data = df)
sns.swarmplot(x= 'tip',y = 'day',data = df)
```

## 🛍️ Real-World Business Analytics on Transaction Data
This module covers end-to-end data visualization techniques applied to real enterprise retail transactions (`sales_data.csv`), analyzing category revenue, regional distributions, and time-series trends using both Matplotlib and Seaborn.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Load enterprise sales dataset
sales_df = pd.read_csv('sales_data.csv')
print(f"Sales Records: {sales_df.shape[0]} rows, {sales_df.shape[1]} columns")
sales_df.head(3)
```

```python
# Inspect data schema and missing values
sales_df.info()
```

### 1. Total Sales by Product Category (Matplotlib Bar Plot)
Aggregating revenue by product category and rendering a clean bar chart with data labels.

```python
# Aggregate total revenue by product category
category_sales = sales_df.groupby('Product Category')['Total Revenue'].sum().sort_values(ascending=False)

plt.figure(figsize=(9, 4.5))
bars = plt.bar(category_sales.index, category_sales.values, color='teal', edgecolor='black', alpha=0.85)
plt.title('Total Revenue by Product Category (Matplotlib)', fontsize=14, fontweight='bold')
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Total Revenue ($)', fontsize=12)
plt.xticks(rotation=20)
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Add value labels on top of bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + (height * 0.01),
             f'${height:,.0f}', ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()
```

### 2. Time-Series Sales Trend Over Time (Matplotlib Line Chart)
Tracking daily revenue trajectory to identify demand cycles and spikes.

```python
# Aggregate revenue by Date
sales_df['Date'] = pd.to_datetime(sales_df['Date'])
daily_trend = sales_df.groupby('Date')['Total Revenue'].sum().reset_index()

plt.figure(figsize=(11, 4.5))
plt.plot(daily_trend['Date'], daily_trend['Total Revenue'], color='crimson', linewidth=2, marker='o', markersize=4)
plt.title('Daily Revenue Trend (Matplotlib Time-Series)', fontsize=14, fontweight='bold')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Daily Revenue ($)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

### 3. Total Sales by Product Category & Region with Seaborn Aggregation
Using Seaborn's `estimator=sum` parameter to compute totals automatically with category grouping.

```python
plt.figure(figsize=(10, 5))
sns.barplot(
    data=sales_df, 
    x='Product Category', 
    y='Total Revenue', 
    estimator=sum, 
    palette='Blues_d', 
    edgecolor='black',
    errorbar=None
)
plt.title('Total Revenue by Product Category (Seaborn)', fontsize=14, fontweight='bold')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=20)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 4.5))
sns.barplot(
    data=sales_df, 
    x='Region', 
    y='Total Revenue', 
    estimator=sum, 
    palette='coolwarm', 
    edgecolor='black',
    errorbar=None
)
plt.title('Total Revenue by Geographic Region (Seaborn)', fontsize=14, fontweight='bold')
plt.ylabel('Total Sales ($)')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### 4. Revenue Distribution across Payment Methods (Seaborn Boxplot)
Evaluating ticket sizes and transaction values across payment channels.

```python
plt.figure(figsize=(8, 4.5))
sns.boxplot(
    data=sales_df, 
    x='Payment Method', 
    y='Total Revenue', 
    palette='Set2'
)
plt.title('Transaction Value Distribution by Payment Method', fontsize=13, fontweight='bold')
plt.ylabel('Transaction Revenue ($)')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### 5. Numerical Feature Correlation Matrix (Seaborn Heatmap)
Analyzing relationships between `Units Sold`, `Unit Price`, and `Total Revenue`.

```python
sales_num_corr = sales_df[['Units Sold', 'Unit Price', 'Total Revenue']].corr()

plt.figure(figsize=(5.5, 4))
sns.heatmap(sales_num_corr, annot=True, cmap='viridis', fmt='.3f', linewidths=1)
plt.title('Correlation Matrix: Sales Numeric Features', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 📊 General Visualization Questions & Solutions

Here we answer essential, high-frequency plotting and data visualization questions using Matplotlib and Seaborn on a standardized student performance dataset.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Student Performance Dataset for Visualization Practice
viz_df = pd.DataFrame({
    'Student': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Hannah'],
    'Age': [18, 19, 17, 18, 20, 19, 18, 20],
    'Math': [85, 92, 76, 65, 90, 88, 72, 95],
    'Science': [78, 88, 95, 70, 85, 82, 68, 98],
    'Attendance': [92, 95, 85, 70, 90, 88, 75, 98],
    'Grade': ['B', 'A', 'B', 'C', 'A', 'A', 'C', 'A']
})

viz_df
```

```python
# 1. Line Plot: Plot Math marks progression across students.
plt.figure(figsize=(8, 4))
plt.plot(viz_df['Student'], viz_df['Math'], marker='o', color='royalblue', linewidth=2, linestyle='-', label='Math Marks')
plt.title('Math Marks Progression', fontsize=14, fontweight='bold')
plt.xlabel('Student', fontsize=12)
plt.ylabel('Marks', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

```python
# 2. Bar Plot: Compare average Math and Science marks by Grade using Seaborn.
melted_scores = viz_df.melt(id_vars=['Grade'], value_vars=['Math', 'Science'], var_name='Subject', value_name='Score')

plt.figure(figsize=(7, 4))
sns.barplot(data=melted_scores, x='Grade', y='Score', hue='Subject', palette='Set2')
plt.title('Average Subject Marks by Grade', fontsize=13, fontweight='bold')
plt.ylabel('Average Score')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
# 3. Scatter Plot: Math vs Science with Attendance mapped to point size and Grade to color.
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=viz_df, 
    x='Math', 
    y='Science', 
    hue='Grade', 
    size='Attendance', 
    sizes=(60, 250), 
    palette='viridis'
)
plt.title('Math vs Science (Size = Attendance %, Color = Grade)', fontsize=13, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# 4. Histogram & KDE: Distribution of Math scores.
plt.figure(figsize=(7, 4))
sns.histplot(viz_df['Math'], kde=True, bins=5, color='darkorange', edgecolor='black')
plt.title('Math Marks Distribution with KDE Overlay', fontsize=13, fontweight='bold')
plt.xlabel('Math Score')
plt.ylabel('Count')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
# 5. Box Plot: Compare score spreads between Math and Science for outlier detection.
plt.figure(figsize=(6, 4))
sns.boxplot(data=viz_df[['Math', 'Science']], palette='pastel')
plt.title('Box Plot of Math vs Science Marks', fontsize=13, fontweight='bold')
plt.ylabel('Marks')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### 1. Object-Oriented API (`fig, ax`) vs State-Machine (`plt.plot`)
**Interview Question:** *Why do production codebases and senior data scientists strongly mandate the Object-Oriented API (`fig, ax = plt.subplots()`) over `pyplot` convenience functions (`plt.plot()`)?*

- **`plt.plot()` (State-Machine):** Emulates MATLAB. It implicitly tracks a "current active figure and axis". Highly error-prone when creating complex dashboards or subplots because commands apply to whichever axis happened to be active last. Not thread-safe!
- **`fig, ax = plt.subplots()` (Object-Oriented):** Directly exposes the `Figure` and `Axes` objects. Allows explicit, modular plotting where functions take an `ax` parameter and draw onto it without global side-effects.

```python
# Demonstrating Modular Function Design with OO API
def plot_distribution(data, feature, ax, color='teal'):
    """Reusable plotting function that accepts an explicit ax object."""
    sns.histplot(data[feature], kde=True, ax=ax, color=color)
    ax.set_title(f'Distribution of {feature}', fontweight='bold')
    ax.grid(axis='y', linestyle='--', alpha=0.5)

# Clean grid assembly:
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
plot_distribution(viz_df, 'Math', ax=axes[0], color='royalblue')
plot_distribution(viz_df, 'Science', ax=axes[1], color='darkorange')
plt.tight_layout()
plt.show()
```

### 2. Seaborn Figure-Level vs. Axes-Level Functions (The Major Seaborn Trap)
**Interview Question:** *Why does `sns.scatterplot(..., ax=ax)` work inside a subplot grid, but `sns.relplot(..., ax=ax)` fails or creates a brand new window?*

| Feature | **Axes-Level Functions** | **Figure-Level Functions** |
|---|---|---|
| **Examples** | `sns.scatterplot`, `sns.histplot`, `sns.boxplot`, `sns.barplot` | `sns.relplot`, `sns.displot`, `sns.catplot`, `sns.jointplot` |
| **Output Type** | Matplotlib `AxesSubplot` | Seaborn `FacetGrid` |
| **Accepts `ax=` parameter?** | **YES** (fits inside existing subplots) | **NO** (manages its own figure and layout) |
| **Use Case** | Custom multi-plot dashboards using `plt.subplots()` | Multi-variable facet grids (`col=...`, `row=...`) |

```python
# Figure-level functions manage their own figure (FacetGrid)
# Ideal for multi-column faceting:
g = sns.catplot(
    data=viz_df, 
    x='Grade', 
    y='Math', 
    kind='box', 
    col='Age', 
    col_wrap=2, 
    height=3, 
    aspect=1.2
)
g.fig.subplots_adjust(top=0.85)
g.fig.suptitle('Math Score across Grades Faceted by Age', fontweight='bold')
plt.show()
```

### 3. Preventing Memory Leaks in Batch / Production Pipelines
**Interview Question:** *Your batch reporting script generates 10,000 plots and crashes with an Out-of-Memory (OOM) error. How do you prevent Matplotlib memory leaks?*

- Matplotlib maintains a global list of all created figures in memory until explicitly closed.
- Calling `plt.show()` does **NOT** free memory!
- **The Solution:**
  1. Always call `plt.close(fig)` or `plt.close('all')` inside a `try...finally` block or loop.
  2. For headless servers (no display), set the non-interactive backend: `matplotlib.use('Agg')` before importing pyplot.

```python
# Safe Plotting Loop Pattern (Zero Memory Leaks)
import matplotlib
# matplotlib.use('Agg') # Set for headless backend in batch scripts

for i in range(3):
    fig, ax = plt.subplots(figsize=(4, 2))
    ax.plot([1, 2, 3], [i, i**2, i**3])
    ax.set_title(f'Iteration {i}')
    # Save to disk:
    # fig.savefig(f'plot_{i}.png', bbox_inches='tight')
    
    # CRITICAL: Always explicitly close the figure!
    plt.close(fig)

print("Batch loop completed safely with 0 leaked figure references!")
```

### 4. Correlation Matrix Heatmap with Upper Triangle Mask (`np.triu`)
**Interview Question:** *Why and how should you mask the upper triangle of a correlation matrix heatmap?*

- A correlation matrix is symmetric ($Corr(A, B) = Corr(B, A)$) and the diagonal is always 1.0. Displaying the full matrix creates cognitive clutter.
- Use `np.triu(np.ones_like(corr, dtype=bool))` to create a boolean mask for the redundant upper triangle.

```python
# Professional Correlation Heatmap
corr = viz_df[['Age', 'Math', 'Science', 'Attendance']].corr()

# Generate mask for upper triangle
mask = np.triu(np.ones_like(corr, dtype=bool))

plt.figure(figsize=(6, 5))
sns.heatmap(
    corr, 
    mask=mask, 
    annot=True, 
    cmap='coolwarm', 
    vmin=-1, 
    vmax=1, 
    fmt='.2f', 
    linewidths=1,
    square=True,
    cbar_kws={"shrink": 0.8}
)
plt.title('Clean Lower-Triangle Correlation Heatmap', fontweight='bold', fontsize=12)
plt.show()
```

### 6. The Executive Chart Chooser Matrix & Cognitive Pitfalls
**Interview Question:** *Why are pie charts strongly discouraged in professional data science presentations? What should you use instead?*

- **Cognitive Flaw of Pie Charts:** Human visual perception struggles to judge non-adjacent 2D angles and arc lengths accurately. Differences of 5–10% between pie slices are virtually indistinguishable.
- **Superior Alternatives:**
  - **Horizontal Bar Charts:** Ranked by magnitude — human eyes discern 1D linear lengths with sub-millimeter precision.
  - **Lollipop Charts:** Less visual ink than solid bars, excellent for ranking 10+ categories.
  - **Treemaps:** When showing hierarchical part-to-whole proportions.

```python
# Demonstration: Bar Chart vs Pie Chart Cognitive Clarity
categories = ['Product A', 'Product B', 'Product C', 'Product D', 'Product E']
market_share = [24, 22, 20, 18, 16] # Subtly different percentages

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

# ❌ Pie Chart (Hard to rank A vs B vs C):
axes[0].pie(market_share, labels=categories, autopct='%1.0f%%', startangle=90, colors=sns.color_palette('pastel'))
axes[0].set_title('❌ Pie Chart: Angle Judgments are Difficult', fontweight='bold')

# ✅ Horizontal Bar Chart (Instantly clear ranking):
y_pos = np.arange(len(categories))
axes[1].barh(y_pos, sorted(market_share), color='steelblue', edgecolor='black')
axes[1].set_yticks(y_pos)
axes[1].set_yticklabels(sorted(categories))
axes[1].set_xlabel('Market Share (%)')
axes[1].set_title('✅ Bar Chart: Linear Lengths are Immediately Readable', fontweight='bold')
axes[1].grid(axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

### 7. Solving Overplotting for Large Datasets (100k+ Points)
**Interview Question:** *How do you visualize a scatter plot with 200,000 data points without turning the plot into an opaque blob?*

**Techniques:**
1. **Alpha blending (`alpha=0.05` to `0.1`):** Shows density where points accumulate.
2. **Hexbin Plots (`plt.hexbin(..., gridsize=30)`):** Bins 2D space into regular hexagons and colors by count. Far superior to standard scatter plots for large data!
3. **2D KDE Contour Overlays:** Maps smooth continuous density contours.

```python
# Overplotting Solutions: Scatter vs Hexbin
np.random.seed(42)
n_large = 50_000
x_big = np.random.randn(n_large)
y_big = 2 * x_big + np.random.randn(n_large) * 1.5

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# ❌ Standard Scatter with Overplotting:
axes[0].scatter(x_big, y_big, s=1, color='blue', alpha=0.1)
axes[0].set_title('Alpha-Blended Scatter (50k points)', fontweight='bold')
axes[0].grid(True, linestyle='--', alpha=0.5)

# ✅ Hexagonal Binning (Hexbin):
hb = axes[1].hexbin(x_big, y_big, gridsize=35, cmap='Blues', mincnt=1)
cb = fig.colorbar(hb, ax=axes[1])
cb.set_label('Point Density')
axes[1].set_title('Hexbin Density Heatmap (Clear Clustered Density)', fontweight='bold')

plt.tight_layout()
plt.show()
```

### 8. Asymmetric Dashboard Layouts with Matplotlib `GridSpec`
**Interview Question:** *How do you build a custom JointPlot (main scatter with marginal top and side distribution histograms) using raw Matplotlib without relying on Seaborn?*

- **`matplotlib.gridspec.GridSpec`:** Provides a flexible grid geometry manager to span subplots across unequal rows and columns.

```python
# Asymmetric Subplots via GridSpec (Custom JointPlot)
from matplotlib.gridspec import GridSpec

fig = plt.figure(figsize=(8, 8))
gs = GridSpec(4, 4, hspace=0.15, wspace=0.15)

# Main Scatter plot (spanning bottom 3 rows, left 3 columns):
ax_main = fig.add_subplot(gs[1:4, 0:3])

# Top Marginal Histogram (spanning top row, left 3 columns):
ax_top = fig.add_subplot(gs[0, 0:3], sharex=ax_main)

# Right Marginal Histogram (spanning bottom 3 rows, right column):
ax_right = fig.add_subplot(gs[1:4, 3], sharey=ax_main)

# Data
np.random.seed(42)
x_data = np.random.normal(loc=100, scale=15, size=500)
y_data = 0.6 * x_data + np.random.normal(loc=20, scale=10, size=500)

# Plot components
ax_main.scatter(x_data, y_data, color='teal', alpha=0.6, s=25)
ax_top.hist(x_data, bins=25, color='teal', edgecolor='black', alpha=0.7)
ax_right.hist(y_data, bins=25, orientation='horizontal', color='teal', edgecolor='black', alpha=0.7)

# Clean marginal axes:
plt.setp(ax_top.get_xticklabels(), visible=False)
plt.setp(ax_right.get_yticklabels(), visible=False)
ax_top.set_ylabel('Count')
ax_right.set_xlabel('Count')
ax_main.set_xlabel('Feature X')
ax_main.set_ylabel('Feature Y')

plt.suptitle('Custom Marginal JointPlot using Matplotlib GridSpec', y=0.92, fontweight='bold')
plt.show()
```

### 9. Color Theory & Accessibility in Visualizations
**Interview Question:** *When should you use Diverging vs Sequential vs Qualitative colormaps? How do you ensure charts are accessible to colorblind viewers?*

- **Sequential (`viridis`, `Blues`, `plasma`):** Ordered data with single direction (0 to 100).
- **Diverging (`coolwarm`, `vlag`, `RdBu`):** Data with a meaningful midpoint or critical zero (e.g. profit/loss, z-scores, net temperature anomaly). Use `vcenter=0` to balance.
- **Qualitative (`Set2`, `tab10`, `colorblind`):** Unordered categorical groups.
- **Accessibility:** Avoid pure Red/Green encoding! Use Seaborn's `palette='colorblind'` or `viridis` which are perceptually uniform in grayscale and all forms of colorblindness.

```python
# Diverging Colormap Centered at Critical Zero Point
from matplotlib.colors import TwoSlopeNorm

data_matrix = np.array([
    [-5.2, -1.8, 3.4],
    [2.1, 0.0, -4.6],
    [4.8, 1.2, -0.3]
]) # Financial net profit/loss

fig, ax = plt.subplots(figsize=(6, 4))
# TwoSlopeNorm guarantees that value 0 is assigned exact neutral white/center color:
norm = TwoSlopeNorm(vmin=-6, vcenter=0, vmax=6)
im = ax.imshow(data_matrix, cmap='RdBu_r', norm=norm)
fig.colorbar(im, ax=ax, label='Profit / Loss ($K)')
ax.set_title('Diverging Colormap Anchored at Neutral 0', fontweight='bold')

for i in range(3):
    for j in range(3):
        ax.text(j, i, f"{data_matrix[i, j]:+.1f}", ha='center', va='center', color='black', fontweight='bold')

plt.show()
```

### 10. Dynamic Axis Tick Formatters (`matplotlib.ticker`)
**Interview Question:** *How do you cleanly format axis numbers as `$1.5M`, `€250K`, or `45%` without converting the underlying numerical data to strings?*

- Converting data to strings breaks numerical scale interpolation and axis limits.
- Use `matplotlib.ticker.FuncFormatter` or `PercentFormatter` to format tick labels on-the-fly dynamically!

```python
# Dynamic Currency Formatting using FuncFormatter
import matplotlib.ticker as ticker

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
revenue = [1_200_000, 2_450_000, 1_890_000, 3_150_000, 2_800_000, 4_100_000]

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(months, revenue, marker='o', linewidth=2, color='navy')
ax.set_title('Monthly Revenue with Dynamic Tick Formatter', fontweight='bold')

# Custom formatter for Millions ($M):
def currency_fmt(val, pos):
    return f"${val * 1e-6:.1f}M"

ax.yaxis.set_major_formatter(ticker.FuncFormatter(currency_fmt))
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### 11. Statistical Inference: Confidence Intervals in Seaborn
**Interview Question:** *What do the shaded bands or error bars in `sns.lineplot` and `sns.barplot` represent by default? How are they calculated?*

- By default, Seaborn calculates a **95% Confidence Interval using Bootstrapping** (resampling data 1,000 times with replacement).
- It represents the estimated uncertainty of the mean estimate, **not** the standard deviation of raw data points!
- To display standard deviation of observations instead: use `errorbar='sd'`.
- To turn off error bars: use `errorbar=None`.

```python
# Confidence Interval (Bootstrapped 95% CI) vs Standard Deviation (SD)
tips = sns.load_dataset('tips')

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

# 1. Default: 95% Confidence Interval (Bootstrapped)
sns.barplot(data=tips, x='day', y='total_bill', ax=axes[0], palette='Blues_d', errorbar=('ci', 95))
axes[0].set_title('Default: 95% Bootstrapped CI of the Mean', fontweight='bold')

# 2. Standard Deviation of the Sample Observations:
sns.barplot(data=tips, x='day', y='total_bill', ax=axes[1], palette='Purples_d', errorbar='sd')
axes[1].set_title("errorbar='sd': Standard Deviation Spread", fontweight='bold')

plt.tight_layout()
plt.show()
```
