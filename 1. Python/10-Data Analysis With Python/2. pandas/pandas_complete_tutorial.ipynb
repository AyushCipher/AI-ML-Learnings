```python
#loading the data
import numpy as np 
import pandas as pd 
```

```python
df = pd.read_csv(r'anime.csv')
```

```python
df.head()
```

```python
df.loc[4]['Title']
```

```python
def extract_episodes(txt):
    check = False
    data = ""
    for i in txt:
        if i == ")":
            check = False
            return data
        if check == True: 
            data = data + i 
        if i == '(':
            check = True 
           
```

```python
df["Episodes"] = df["Title"].apply(extract_episodes)
```

```python
df['Episodes']= df['Episodes'].str.replace(" eps","")
```

```python
df['Episodes'] = df['Episodes'].astype(int)
```

```python
df
```

### Feature Extraction Analytical Questions & Solutions
Below we solve all feature extraction and dataset ranking questions on the anime dataset:
1. Extract episode counts from titles
2. Extract start and end airing time stamps
3. Calculate total duration in months
4. Identify highest scoring anime
5. Find top 5 highest scoring anime
6. Find highest episode count anime
7. Find top 5 highest episode counts
8. Identify the longest running anime

```python
df.loc[3]['Title']
```

```python
def extraction_time(txt):
    check = False
    data = ""
    for i in range(len(txt)):
        if txt[i] == ')':
            for j in range(i+1,i+20):
                data += txt[j]

            return data
            
            
```

```python
df['Total Time'] = df['Title'].apply(extraction_time)
```

```python
df.head()
```

```python
from dateutil.relativedelta import relativedelta
from datetime import datetime

def calculate_total_months(period):
    try:
        start_str, end_str = period.split(' - ')
        start_date = datetime.strptime(start_str, '%b %Y')
        end_date = datetime.strptime(end_str, '%b %Y')
        r = relativedelta(end_date, start_date)
        return r.years * 12 + r.months + 1  # +1 to include the starting month
    except:
        return None

df['Months'] = df['Total Time'].apply(calculate_total_months)
```

```python
df
```

```python
# 1. Highest scoring anime
print("--- Highest Scoring Anime ---")
highest_score_anime = df[df['Score'] == df['Score'].max()][['Title', 'Score']]
display(highest_score_anime)

# 2. Top 5 highest scoring anime
print("\n--- Top 5 Highest Scoring Anime ---")
top5_score = df.sort_values(by='Score', ascending=False)[['Title', 'Score']].head(5)
display(top5_score)

# 3. Anime with the highest episode count
print("\n--- Anime with Highest Episode Count ---")
highest_episodes_anime = df[df['Episodes'] == df['Episodes'].max()][['Title', 'Episodes']]
display(highest_episodes_anime)

# 4. Top 5 animes with highest episode count
print("\n--- Top 5 Animes with Highest Episode Counts ---")
top5_episodes = df.sort_values(by='Episodes', ascending=False)[['Title', 'Episodes']].head(5)
display(top5_episodes)

# 5. Longest running anime (in months)
print("\n--- Longest Running Anime (in Months) ---")
longest_running = df[df['Months'] == df['Months'].max()][['Title', 'Total Time', 'Months']]
display(longest_running)
```

### 1. Memory Optimization & Dtype Downcasting
**Interview Question:** *Your Pandas DataFrame takes 8 GB of RAM and crashes the server. How do you reduce its memory usage by 70–90% without losing data?*

**Techniques:**
1. **Categorical conversion:** String columns with low cardinality (repeated values like cities, genders, status) store repeated pointer objects. Converting to `'category'` replaces strings with lightweight integer codes and an internal lookup dictionary.
2. **Numeric downcasting:** `int64` (8 bytes) to `int8` (1 byte) or `int16` (2 bytes); `float64` to `float32`.
3. **Inspect memory:** Use `df.info(memory_usage='deep')` and `df.memory_usage(deep=True)`. Regular `.memory_usage()` misses the heap size of Python strings!

```python
# Memory Optimization Demo
import pandas as pd
import numpy as np

# Create sample dataset with high repetition strings
n_rows = 100_000
df_mem = pd.DataFrame({
    'user_id': np.arange(n_rows),
    'membership': np.random.choice(['Bronze', 'Silver', 'Gold', 'Platinum'], size=n_rows),
    'city': np.random.choice(['New York', 'London', 'Tokyo', 'Berlin', 'Mumbai'], size=n_rows),
    'age': np.random.randint(18, 70, size=n_rows),
    'score': np.random.uniform(50.0, 100.0, size=n_rows)
})

mem_before = df_mem.memory_usage(deep=True).sum() / (1024**2)
print(f"Memory BEFORE optimization: {mem_before:.2f} MB")

# 1. Convert low-cardinality strings to 'category'
df_mem['membership'] = df_mem['membership'].astype('category')
df_mem['city'] = df_mem['city'].astype('category')

# 2. Downcast numeric columns
df_mem['age'] = pd.to_numeric(df_mem['age'], downcast='integer') # int8 (18-70 fits in int8)
df_mem['score'] = pd.to_numeric(df_mem['score'], downcast='float') # float32

mem_after = df_mem.memory_usage(deep=True).sum() / (1024**2)
print(f"Memory AFTER optimization:  {mem_after:.2f} MB")
print(f"Memory Reduction: {(1 - mem_after/mem_before)*100:.1f}%")
```

### 2. The Dreaded `SettingWithCopyWarning` Explained
**Interview Question:** *Why does `SettingWithCopyWarning` occur in Pandas, and how do you resolve it? What is chained indexing?*

- **Chained Indexing:** `df['colA'][df['colB'] > 5] = 100` involves two separate operations:
  1. `df['colA']` -> `__getitem__`
  2. `[df['colB'] > 5] = 100` -> `__setitem__`
- Because Pandas cannot guarantee whether the first step returned a **view** or a **copy**, modifying it may fail silently or corrupt state.
- **The Fix:**
  - If modifying original: Use single-indexer `.loc[row_mask, col_mask] = value`.
  - If creating a standalone sub-frame: Explicitly do `.copy()`.

```python
# SettingWithCopyWarning Demo and Fix
df_demo = pd.DataFrame({
    'department': ['IT', 'HR', 'IT', 'Sales'],
    'salary': [50000, 45000, 60000, 52000]
})

# ❌ WRONG (Chained Indexing - may trigger SettingWithCopyWarning):
# sub_df = df_demo[df_demo['department'] == 'IT']
# sub_df['salary'] = 70000

# ✅ CORRECT Solution 1: Use .loc directly on original df
df_demo.loc[df_demo['department'] == 'IT', 'salary'] += 5000
print("Updated original dataframe with .loc:")
print(df_demo)

# ✅ CORRECT Solution 2: Explicit .copy() when creating a subset
it_team = df_demo[df_demo['department'] == 'IT'].copy()
it_team['bonus'] = 10000 # Completely safe, no warning!
print("
Independent subset with .copy():")
print(it_team)
```

### 3. GroupBy Deep Dive: `agg()` vs `transform()` vs `filter()` vs `apply()`
**Interview Question:** *What is the difference between `agg()`, `transform()`, `filter()`, and `apply()` in Pandas GroupBy? When do you use each?*

| Method | Output Shape | Primary Use Case |
|---|---|---|
| **`agg()`** | Collapses to **1 row per group** | Summary metrics (sum, mean, min, max, count). |
| **`transform()`** | Returns **same shape as input** | Calculating group-normalized metrics (e.g., z-scores, % of department budget, imputing group mean). |
| **`filter()`** | Subset of original rows | Dropping entire groups that fail an aggregate condition (e.g. keep departments with > 5 employees). |
| **`apply()`** | Flexible (any shape) | Complex custom logic across sub-DataFrames (slower, use vectorized methods when possible). |

```python
# GroupBy: agg vs transform vs filter vs apply
sales_df = pd.DataFrame({
    'Store': ['A', 'A', 'A', 'B', 'B', 'C'],
    'Product': ['Apple', 'Banana', 'Orange', 'Apple', 'Banana', 'Apple'],
    'Revenue': [100, 200, 300, 400, 500, 50]
})
print("Original Sales Data:
", sales_df)

# 1. agg: Summarizes to one row per store
print("
1. agg() - Total Revenue per Store:")
print(sales_df.groupby('Store')['Revenue'].agg(['sum', 'mean']))

# 2. transform: Computes group sum and broadcasts to original row size!
sales_df['Store_Total'] = sales_df.groupby('Store')['Revenue'].transform('sum')
sales_df['Pct_Of_Store'] = (sales_df['Revenue'] / sales_df['Store_Total']) * 100
print("
2. transform() - % Contribution to Store Total:
", sales_df)

# 3. filter: Keep only stores with total revenue >= 500
filtered_stores = sales_df.groupby('Store').filter(lambda g: g['Revenue'].sum() >= 500)
print("
3. filter() - Stores with Revenue >= 500:
", filtered_stores[['Store', 'Product', 'Revenue']])
```

### 5. Window Functions: Rolling, Expanding, and Lead/Lag (`shift`, `pct_change`)
**Interview Question:** *How do you compute a 3-day moving average, cumulative revenue, and day-over-day growth rate for stock or sales data in Pandas?*

- **`.rolling(window=N)`:** Moving window over past $N$ observations.
- **`.expanding()`:** Cumulative window including all observations from the beginning up to the current row.
- **`.shift(1)`:** Lag feature (yesterday's value).
- **`.pct_change()`:** Percentage change $(x_t - x_{t-1}) / x_{t-1}$.

```python
# Window and Time-Series Metrics
dates = pd.date_range(start='2026-01-01', periods=7, freq='D')
stock_df = pd.DataFrame({
    'Date': dates,
    'Price': [100.0, 105.0, 102.0, 110.0, 115.0, 112.0, 120.0]
}).set_index('Date')

# 3-Day Moving Average
stock_df['MA_3D'] = stock_df['Price'].rolling(window=3).mean()

# Cumulative Max and Running Total
stock_df['Running_Max'] = stock_df['Price'].expanding().max()

# Previous Day Price (Lag) and Day-over-Day Return
stock_df['Prev_Price'] = stock_df['Price'].shift(1)
stock_df['Daily_Return_Pct'] = stock_df['Price'].pct_change() * 100

print("Stock Metrics with Window Functions:")
print(stock_df)
```

### 6. Reshaping Masterclass: `melt()` vs `pivot_table()` vs `explode()`
**Interview Question:** *What is the difference between Wide and Long data formats? How do you unpivot wide data, pivot long data, and unnest list elements in cells?*

- **`melt()` (Wide -> Long):** Turns multiple column headers into rows. Ideal for machine learning and tidy datasets.
- **`pivot_table()` (Long -> Wide):** Aggregates and reshapes long data into a matrix. Handles duplicate index/column pairs with an aggregation function.

### 7. Safe Merging: Validation (`validate='1:m'`) & Merge Indicator
**Interview Question:** *How do you prevent accidental Cartesian products / duplicated rows during joins in Pandas? How do you find non-matching rows in an outer join?*

- **`validate` parameter:** Validates merge integrity (`'1:1'`, `'1:m'`, `'m:1'`, `'m:m'`). Raises `MergeError` if duplicates violate your assumption.
- **`indicator=True`:** Adds `_merge` column with `'left_only'`, `'right_only'`, or `'both'`.

```python
# Safe Merging with validate and indicator
users = pd.DataFrame({'user_id': [1, 2, 3], 'name': ['Alice', 'Bob', 'Charlie']})
orders = pd.DataFrame({'user_id': [1, 1, 2, 4], 'amount': [250, 150, 400, 300]})

# Merge with validate and indicator
merged_df = pd.merge(
    users, 
    orders, 
    on='user_id', 
    how='outer',
    validate='1:m',      # Ensures user_id is unique on left table!
    indicator=True       # Shows match source
)
print("Safe Merge Result with indicator:")
print(merged_df)

# Find records that only existed in orders (unregistered users):
orphan_orders = merged_df[merged_df['_merge'] == 'right_only']
print("
Orders without matching User record:")
print(orphan_orders)
```

### 8. Clean Method Chaining (Production-Grade Pandas Pipelines)
**Interview Question:** *How do you write readable, production-grade Pandas transformations without creating 10 intermediate DataFrame variables?*

Use **Method Chaining** with `.assign()`, `.query()`, and `.pipe()`:
- `.assign()`: Adds or modifies columns inline without mutating original data.
- `.query()`: Filters rows using concise SQL-like expressions.
- `.pipe()`: Passes the DataFrame cleanly into external custom functions.

```python
# Method Chaining Pipeline
raw_data = pd.DataFrame({
    'customer': ['Cust_A', 'Cust_B', 'Cust_C', 'Cust_D', 'Cust_E'],
    'spend': [1200, 450, 3100, 800, 150],
    'country': ['US', 'CA', 'US', 'UK', 'US']
})

def flag_tier(df):
    """Custom business logic function used with pipe()"""
    df['is_vip'] = df['spend'] > 1000
    return df

# Fluent pipeline:
clean_summary = (
    raw_data
    .query("country in ['US', 'UK']")
    .assign(tax=lambda d: d['spend'] * 0.08)
    .assign(total=lambda d: d['spend'] + d['tax'])
    .pipe(flag_tier)
    .sort_values(by='total', ascending=False)
    .reset_index(drop=True)
)

print("Final Pipeline Output:")
print(clean_summary)
```

### 10. SQL Window Functions in Pandas (`ROW_NUMBER`, `RANK`, `DENSE_RANK`, `NTILE`)
**Interview Question:** *How do you write the exact equivalents of SQL's `ROW_NUMBER()`, `RANK()`, and `DENSE_RANK() OVER (PARTITION BY Dept ORDER BY Salary DESC)` in Pandas?*

- **`ROW_NUMBER()`:** `df.groupby('Dept').cumcount() + 1`
- **`RANK()`:** `df.groupby('Dept')['Salary'].rank(method='min', ascending=False)` (leaves gaps after ties)
- **`DENSE_RANK()`:** `df.groupby('Dept')['Salary'].rank(method='dense', ascending=False)` (no gaps after ties)
- **`NTILE(N)`:** `pd.qcut(df['Salary'], q=N, labels=False) + 1`

```python
# SQL Window Functions in Pandas
df_sql = pd.DataFrame({
    'Department': ['IT', 'IT', 'IT', 'IT', 'Sales', 'Sales', 'Sales'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace'],
    'Salary': [120000, 120000, 100000, 90000, 110000, 110000, 85000]
})

# Sort within partitions first:
df_sql = df_sql.sort_values(by=['Department', 'Salary'], ascending=[True, False]).reset_index(drop=True)

# 1. ROW_NUMBER()
df_sql['row_number'] = df_sql.groupby('Department').cumcount() + 1

# 2. RANK() - ties get same rank, skips next rank
df_sql['rank'] = df_sql.groupby('Department')['Salary'].rank(method='min', ascending=False).astype(int)

# 3. DENSE_RANK() - ties get same rank, does NOT skip next rank
df_sql['dense_rank'] = df_sql.groupby('Department')['Salary'].rank(method='dense', ascending=False).astype(int)

print("SQL Window Ranking Equivalents in Pandas:")
display(df_sql)
```

### 14. Classic Senior Interview Problem: Customer Retention & Cohort Analysis
**Interview Question:** *Given a transaction table `(customer_id, order_date, amount)`, calculate month-over-month cohort retention rates.*

**Algorithm:**
1. Determine each customer's first purchase month (Cohort Month).
2. Compute the month offset between transaction month and Cohort Month.
3. Group by `[Cohort_Month, Month_Offset]` and count unique active customers.
4. Pivot into a retention matrix and divide by initial cohort size.

```python
# Cohort Retention Matrix
np.random.seed(42)
n_tx = 1000
tx_df = pd.DataFrame({
    'customer_id': np.random.randint(100, 200, size=n_tx),
    'order_date': pd.to_datetime('2026-01-01') + pd.to_timedelta(np.random.randint(0, 90, size=n_tx), unit='D'),
    'amount': np.random.uniform(20, 200, size=n_tx)
})

# 1. Invoice Month & Cohort Month
tx_df['order_month'] = tx_df['order_date'].dt.to_period('M')
tx_df['cohort_month'] = tx_df.groupby('customer_id')['order_month'].transform('min')

# 2. Cohort Index (Month Offset)
tx_df['cohort_index'] = (tx_df['order_month'].dt.year - tx_df['cohort_month'].dt.year) * 12 + \
                        (tx_df['order_month'].dt.month - tx_df['cohort_month'].dt.month)

# 3. Pivot retention table
cohort_data = tx_df.groupby(['cohort_month', 'cohort_index'])['customer_id'].nunique().reset_index()
cohort_pivot = cohort_data.pivot(index='cohort_month', columns='cohort_index', values='customer_id')

# 4. Retention Percentage Matrix
cohort_sizes = cohort_pivot.iloc[:, 0]
retention_matrix = cohort_pivot.divide(cohort_sizes, axis=0) * 100

print("Cohort Retention Rate (%):")
display(retention_matrix.round(1))
```
