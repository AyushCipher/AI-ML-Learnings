#### Abstraction
Abstraction is the concept of hiding the complex implementation details and showing only the necessary features of an object. This helps in reducing programming complexity and effort.

```python
from abc import ABC, abstractmethod

## Abstract base class
class Vehicle(ABC):
    def drive(self):
        print("The vehicle is used for driving")

    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        print("Car engine started")

def operate_vehicle(vehicle):
    vehicle.start_engine()
    vehicle.drive()

car = Car()
operate_vehicle(car)

```

### 1. Abstract Base Classes (`abc.ABC` & `@abstractmethod`)
Abstraction hides complex implementation details and exposes only the essential interface. In Python, abstraction is enforced using the `abc` (Abstract Base Classes) module.

**Key Rules:**
1. A class that inherits from `ABC` and contains at least one `@abstractmethod` **cannot be instantiated**.
2. Any subclass **must implement all abstract methods**; otherwise, instantiating the subclass will raise a `TypeError`.

```python
from abc import ABC, abstractmethod

# Abstract Base Class definition
class Shape(ABC):
    @abstractmethod
    def area(self):
        """Calculate area of the shape"""
        pass
        
    @abstractmethod
    def perimeter(self):
        """Calculate perimeter of the shape"""
        pass

# Attempting to instantiate an abstract class directly:
try:
    s = Shape()
except TypeError as e:
    print("Direct instantiation blocked:", e)
```

### 2. Concrete Implementations of Abstract Base Class
Every child class is strictly required to implement all abstract methods, ensuring guaranteed interface consistency.

```python
import math

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
        
    def area(self):
        return math.pi * (self.radius ** 2)
        
    def perimeter(self):
        return 2 * math.pi * self.radius

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
        
    def area(self):
        return self.width * self.height
        
    def perimeter(self):
        return 2 * (self.width + self.height)

shapes = [Circle(5), Rectangle(4, 6)]
for shape in shapes:
    print(f"{shape.__class__.__name__} -> Area: {shape.area():.2f}, Perimeter: {shape.perimeter():.2f}")
```

### 3. Abstract Properties (`@property` + `@abstractmethod`)
You can mandate that subclasses implement specific properties as well as methods!

```python
class DatabaseClient(ABC):
    @property
    @abstractmethod
    def connection_string(self):
        pass
        
    @abstractmethod
    def connect(self):
        pass

class PostgresClient(DatabaseClient):
    def __init__(self, host, port):
        self.host = host
        self.port = port
        
    @property
    def connection_string(self):
        return f"postgresql://{self.host}:{self.port}/mydb"
        
    def connect(self):
        return f"Connected to Postgres at {self.connection_string}"

pg = PostgresClient("localhost", 5432)
print(pg.connect())
```

### 4. Real-World Design Pattern: Payment Gateway Interface
**Interview Problem:** *Design an enterprise payment processing system that enforces a clean interface across Stripe, PayPal, and Crypto gateways.*

```python
class PaymentGateway(ABC):
    @abstractmethod
    def authenticate(self) -> bool:
        pass
        
    @abstractmethod
    def pay(self, amount: float) -> str:
        pass

class StripeGateway(PaymentGateway):
    def __init__(self, api_key):
        self.api_key = api_key
        
    def authenticate(self) -> bool:
        return len(self.api_key) > 10
        
    def pay(self, amount: float) -> str:
        if self.authenticate():
            return f"✅ Stripe processed ${amount:,.2f}"
        return "❌ Stripe authentication failed"

class PayPalGateway(PaymentGateway):
    def __init__(self, email):
        self.email = email
        
    def authenticate(self) -> bool:
        return "@" in self.email
        
    def pay(self, amount: float) -> str:
        if self.authenticate():
            return f"✅ PayPal processed ${amount:,.2f} for {self.email}"
        return "❌ PayPal authentication failed"

# Client processor function: decoupled from concrete payment providers!
def process_checkout(gateway: PaymentGateway, total: float):
    print(gateway.pay(total))

process_checkout(StripeGateway("sk_live_99887766554433"), 149.99)
process_checkout(PayPalGateway("user@example.com"), 79.50)
```
