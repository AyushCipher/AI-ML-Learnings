#### Inheritance In Python
Inheritance is a fundamental concept in Object-Oriented Programming (OOP) that allows a class to inherit attributes and methods from another class. This lesson covers single inheritance and multiple inheritance, demonstrating how to create and use them in Python.

````python
## Inheritance (Single Inheritance)
## Parent class
class Car:
    def __init__(self, windows, doors, engine_type):
        self.windows = windows
        self.doors = doors
        self.engine_type = engine_type
    
    def drive(self):
        print(f"The person will drive the {self.engine_type} car. ")
````

````python
car1 = Car(4, 5, "petrol")
car1.drive()
````

````python
class Tesla(Car):
    def __init__(self, windows, doors, engine_type, is_self_driving):
        super().__init__(windows, doors, engine_type)
        self.is_self_driving = is_self_driving  # ← renamed

    def self_driving(self):
        print(f"Tesla supports self driving : {self.is_self_driving}")

tesla1 = Tesla(4, 5, "electric", True)
tesla1.self_driving()   
````

````python
tesla1.drive()
````

````python
### Multiple Inheritance
## When a class inherits from more than one base class.
## Base class 1
class Animal:
    def __init__(self,name):
        self.name = name

    def speak(self):
        print("Subclass must implement this method")

## Base class 2
class Pet:
    def __init__(self, owner):
        self.owner = owner


## Derived class
class Dog(Animal,Pet):
    def __init__(self,name,owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)

    def speak(self):
        return f"{self.name} say woof"
    

## Create an object
dog = Dog("Buddy","John")
print(dog.speak())
print(f"Owner:{dog.owner}")


````

#### Conclusion
Inheritance is a powerful feature in OOP that allows for code reuse and the creation of a more logical class structure. Single inheritance involves one base class, while multiple inheritance involves more than one base class. Understanding how to implement and use inheritance in Python will enable you to design more efficient and maintainable object-oriented programs.

### 1. The Diamond Problem & Method Resolution Order (MRO)
**Interview Question:** *What is the Diamond Problem in multiple inheritance? How does Python solve it using C3 Linearization?*

```
     A
   /   \
  B     C
   \   /
     D
```
- In languages like C++, diamond inheritance can cause duplicated base class instances or ambiguous calls.
- Python uses **C3 Linearization** to produce a deterministic, monotonic method resolution order.
- You can inspect the resolution order using `ClassName.mro()` or `ClassName.__mro__`.

````python
# Demonstrating the Diamond Problem and Cooperative super()
class A:
    def process(self):
        print("-> Enter A")
        print("<- Exit A")

class B(A):
    def process(self):
        print("-> Enter B")
        super().process()
        print("<- Exit B")

class C(A):
    def process(self):
        print("-> Enter C")
        super().process()
        print("<- Exit C")

class D(B, C):
    def process(self):
        print("-> Enter D")
        super().process()
        print("<- Exit D")

print("MRO for Class D:")
for idx, cls in enumerate(D.mro(), 1):
    print(f" {idx}. {cls.__name__}")

print("\nExecution Trace of D.process():")
d = D()
d.process()
````

### 2. Composition vs. Inheritance ("HAS-A" vs. "IS-A")
**Interview Question:** *Why do software architects recommend "Favor composition over inheritance"? Provide a concrete example.*

- **Inheritance (IS-A):** Tightly couples child classes to parent internals. Modifying a base class can break children across the codebase (fragile base class problem).
- **Composition (HAS-A):** Composes complex behavior by containing small, interchangeable component objects. Allows swapping behaviors dynamically at runtime!

````python
# Composition (HAS-A) vs Inheritance (IS-A)
# Component 1: Engine behavior
class ElectricEngine:
    def start(self): return "Silent hum... Electric engine running."

class V8Engine:
    def start(self): return "VROOOM! Petrol V8 engine roaring."

# Vehicle composes an Engine rather than inheriting from it:
class Car:
    def __init__(self, model, engine):
        self.model = model
        self.engine = engine  # HAS-A relationship
        
    def drive(self):
        print(f"Driving {self.model}: {self.engine.start()}")

tesla = Car("Model S", ElectricEngine())
mustang = Car("Mustang GT", V8Engine())

tesla.drive()
mustang.drive()
````

### 3. Type Checking: `isinstance()` vs. `type()`
**Interview Question:** *Why is `if type(x) == Parent:` considered an anti-pattern in object-oriented Python?*

- `type(x) == Parent` checks for an **exact class match** and completely ignores inheritance (evaluates to `False` for subclasses).
- `isinstance(x, Parent)` respects the inheritance hierarchy and works polymorphically.

````python
# isinstance() vs type() check
class Animal: pass
class Dog(Animal): pass

dog = Dog()

print("type(dog) == Animal:      ", type(dog) == Animal)       # ❌ False! Breaks polymorphism
print("isinstance(dog, Animal):  ", isinstance(dog, Animal))   # ✅ True! Respects inheritance
print("issubclass(Dog, Animal):  ", issubclass(Dog, Animal))   # ✅ True!
````
