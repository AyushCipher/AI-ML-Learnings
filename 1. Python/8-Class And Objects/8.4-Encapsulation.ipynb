#### Encapsulation And Abstraction
Encapsulation and abstraction are two fundamental principles of Object-Oriented Programming (OOP) that help in designing robust, maintainable, and reusable code. Encapsulation involves bundling data and methods that operate on the data within a single unit, while abstraction involves hiding complex implementation details and exposing only the necessary features.

##### Encapsulation
Encapsulation is the concept of wrapping data (variables) and methods (functions) together as a single unit. It restricts direct access to some of the object's components, which is a means of preventing accidental interference and misuse of the data.


```python
### Encapsulation with Getter and Setter Methods
### Public, protected, private variables or access modifiers

class Person:
    def __init__(self, name, age):
        self.name = name    ## public variables
        self.age = age      ## public variables

def get_name(person):
    return person.name

person = Person("Ayush",21)
get_name(person)
```

```python
dir(person)
```

```python
class Person:
    def __init__(self, name, age, gender):
        self.__name = name      # private variables
        self.__age = age        # private variables
        self.gender = gender

    def get_name(self):
        return self.__name

person = Person("Ayush", 21, "Male")
print(person.get_name())  # "Ayush"
```

```python
dir(person)
```

```python
class Person:
    def __init__(self, name, age, gender):
        self._name = name    ## protected variables
        self._age = age      ## protected variables
        self.gender = gender

class Employee(Person):
    def __init__(self, name, age, gender):
        super().__init__(name, age, gender)



employee = Employee("Ayush",21,"Male")
print(employee._name)
```

```python
## Encapsulation With Getter And Setter
class Person:
    def __init__(self, name, age):
        self.__name = name  ## Private access modifier or variable
        self.__age = age    ## Private variable

    ## getter method for name
    def get_name(self):
        return self.__name
    
    ## setter method for name
    def set_name(self,name):
        self.__name = name

    # Getter method for age
    def get_age(self):
        return self.__age
    
    # Setter method for age
    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Age cannot be negative.")


person = Person("Ayush",21)

## Access and modify private variables using getter and setter

print(person.get_name())
print(person.get_age())

person.set_age(25)
print(person.get_age())

person.set_age(-5)
    
```

### 1. Name Mangling & The "Consenting Adults" Philosophy
**Interview Question:** *How does Python enforce private variables with double underscores `__var`? Can private attributes still be accessed from outside?*

- Python transforms `self.__attribute` into `self._ClassName__attribute`. This is called **Name Mangling**.
- Its purpose is to prevent name collisions in subclasses, **not** to build an impenetrable security barrier.
- Python operates on the principle: *"We are all consenting adults here"*. You can still access `obj._ClassName__var` directly if needed.

```python
# Name Mangling Demystified
class SecretVault:
    def __init__(self, secret_code):
        self.__secret_code = secret_code  # Private attribute

vault = SecretVault("TOP_SECRET_123")

# Direct access fails with AttributeError:
try:
    print(vault.__secret_code)
except AttributeError as e:
    print("Direct access failed:", e)

# But it is accessible via the mangled name!
print("Accessed via mangled name:", vault._SecretVault__secret_code)
```

### 2. Pythonic Properties (`@property`, `@setter`, `@deleter`)
**Interview Question:** *Why is `@property` preferred over Java-style `get_value()` and `set_value()` methods in Python?*

- It allows exposing attributes with clean dot-syntax (`person.age = 25`) while retaining full control over data validation, lazy computation, and read-only protections behind the scenes.

```python
# Production-Grade Encapsulation with @property
class BankAccount:
    def __init__(self, owner, initial_balance=0.0):
        self.owner = owner
        self._balance = float(initial_balance)
        
    # Read-only property
    @property
    def balance(self):
        """Current balance (read-only)"""
        return self._balance
        
    # Validated deposit
    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")
        self._balance += amount
        print(f"Deposited ${amount:,.2f}. New Balance: ${self._balance:,.2f}")
        
    # Validated withdrawal
    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        if amount > self._balance:
            raise ValueError("Insufficient funds!")
        self._balance -= amount
        print(f"Withdrew ${amount:,.2f}. Remaining Balance: ${self._balance:,.2f}")

account = BankAccount("Ayush", 500)
print(f"Owner: {account.owner}, Balance: ${account.balance:,.2f}")

account.deposit(200)
account.withdraw(150)

# Trying to set balance directly without validation fails:
try:
    account.balance = 1000000  # AttributeError: can't set attribute
except AttributeError as e:
    print("Cannot bypass validation:", e)
```
