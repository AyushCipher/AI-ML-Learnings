```python
## Define a class with instance methods
class Dog:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    
    def bark(self):
        print(f"{self.name} says woof")


dog1 = Dog("Buddy", 3)
dog1.bark()
dog2 = Dog("Lucy", 4)
dog2.bark()

```

```python
### Modeling a Bank Account

## Define a class for bank account
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self,amount):
        self.balance += amount
        print(f"{amount} is deposited. New balance is {self.balance}")

    def withdraw(self,amount):
        if amount > self.balance:
            print("Insufficient funds!")
        else:
            self.balance -= amount
            print(f"{amount} is withdrawn. New Balance is {self.balance}")

    def get_balance(self):
        return self.balance
    
## Create an account
account = BankAccount("Ayush",5000)
print(account.balance)

    
```

```python
## Call instance methods
account.deposit(100)
```

```python
account.withdraw(300)
```

```python
print(account.get_balance())
```

### 1. Class Attributes vs. Instance Attributes (The Mutable Attribute Trap)
**Interview Question:** *What is the difference between a class attribute and an instance attribute? What bug happens when a class attribute is a mutable object like a list or dict?*

- **Class Attribute:** Belongs to the class itself, shared across **all** instances.
- **Instance Attribute:** Bound to a specific instance (`self.attr`).
- **The Trap:** If a class attribute is a list (`tricks = []`), modifying it via one instance mutates the shared list for **every** instance!

```python
# The Mutable Class Attribute Bug and the Fix
class BuggyDog:
    tricks = []  # ❌ SHARED mutable class attribute!
    def __init__(self, name):
        self.name = name

dog1 = BuggyDog("Buddy")
dog2 = BuggyDog("Max")
dog1.tricks.append("Roll over")
print("Buggy dog2 tricks (unexpectedly has Buddy's trick!):", dog2.tricks)

# ✅ CORRECT: Instance attribute inside __init__
class SafeDog:
    species = "Canis familiaris"  # Immutable class attribute is fine
    def __init__(self, name):
        self.name = name
        self.tricks = []          # Independent per instance!

safe1 = SafeDog("Buddy")
safe2 = SafeDog("Max")
safe1.tricks.append("Roll over")
print("Safe dog2 tricks (empty as expected):", safe2.tricks)
```

### 2. Method Types: Instance vs. `@classmethod` vs. `@staticmethod`
**Interview Question:** *What are the exact differences between an instance method, a class method, and a static method? When should you use each?*

| Method Type | First Parameter | Decorator | Primary Use Case |
|---|---|---|---|
| **Instance Method** | `self` (the instance) | None | Accessing/mutating instance state. |
| **Class Method** | `cls` (the class) | `@classmethod` | Alternative constructors, class-level factories. |
| **Static Method** | None (pure function) | `@staticmethod` | Utility functions logically grouped inside the class namespace. |

```python
# Instance vs Class vs Static Methods & Alternative Constructors
from datetime import date

class Employee:
    standard_bonus = 1000
    
    def __init__(self, name, birth_year, salary):
        self.name = name
        self.birth_year = birth_year
        self.salary = salary
        
    # 1. Instance method: Uses self
    def calculate_total_pay(self):
        return self.salary + self.standard_bonus
        
    # 2. Class method: Uses cls -> Perfect for alternative constructors!
    @classmethod
    def from_birth_year(cls, name, birth_year, salary):
        return cls(name, birth_year, salary)
        
    @classmethod
    def from_csv_string(cls, employee_csv):
        # Parses "John,1995,75000"
        name, year, sal = employee_csv.split(",")
        return cls(name, int(year), float(sal))
        
    # 3. Static method: Pure utility, does not touch self or cls
    @staticmethod
    def is_working_age(birth_year):
        age = date.today().year - birth_year
        return 18 <= age <= 65

# Creating instances via alternative constructors:
emp1 = Employee.from_csv_string("Alice,1992,95000")
print(f"Created: {emp1.name}, Salary: ${emp1.salary:,.0f}, Total: ${emp1.calculate_total_pay():,.0f}")
print("Is Alice of working age?", Employee.is_working_age(emp1.birth_year))
```

### 3. Object Creation: `__new__` vs. `__init__` (Singleton Pattern)
**Interview Question:** *What is the difference between `__new__` and `__init__`? How do you implement a thread-safe Singleton in Python?*

- **`__new__(cls, ...)`:** The actual constructor. Responsible for allocating and returning a new object instance in memory.
- **`__init__(self, ...)`:** The initializer. Receives the newly allocated instance from `__new__` and populates its attributes.
- **Singleton:** Overriding `__new__` allows returning an existing instance instead of creating a new one.

```python
# Singleton Pattern using __new__
class DatabaseConnection:
    _instance = None
    
    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            print("Allocating NEW DatabaseConnection instance in memory...")
            cls._instance = super().__new__(cls)
            cls._instance._initialized = False
        return cls._instance
        
    def __init__(self, db_name="ProductionDB"):
        if not self._initialized:
            self.db_name = db_name
            self._initialized = True

db1 = DatabaseConnection()
db2 = DatabaseConnection()

print("Are db1 and db2 the exact same instance in memory?", db1 is db2)
```

### 4. Memory Optimization with `__slots__`
**Interview Question:** *What does `__slots__` do, and how does it save 40-50% memory when creating millions of objects?*

- By default, Python objects store their instance attributes in an internal dynamic dictionary (`__dict__`).
- `__slots__` tells Python not to create a `__dict__`, instead allocating a compact fixed-size C array for the specified attribute names.
- **Trade-off:** Prevents adding arbitrary new attributes at runtime.

```python
# Memory Optimization using __slots__
import sys

class RegularPoint:
    def __init__(self, x, y):
        self.x = x
        self.y = y

class SlottedPoint:
    __slots__ = ('x', 'y')  # No __dict__ created!
    def __init__(self, x, y):
        self.x = x
        self.y = y

p_regular = RegularPoint(10, 20)
p_slotted = SlottedPoint(10, 20)

print("Regular instance has __dict__:", hasattr(p_regular, '__dict__'))
print("Slotted instance has __dict__: ", hasattr(p_slotted, '__dict__'))
print("Memory of RegularPoint + dict:", sys.getsizeof(p_regular) + sys.getsizeof(p_regular.__dict__), "bytes")
print("Memory of SlottedPoint:       ", sys.getsizeof(p_slotted), "bytes")
```
