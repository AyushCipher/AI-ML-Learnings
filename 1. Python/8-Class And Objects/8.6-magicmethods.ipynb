### 1. `__str__` vs. `__repr__` Explained
**Interview Question:** *What is the difference between `__str__` and `__repr__`? What happens if `__str__` is not defined?*

- **`__repr__`:** For **developers**. Unambiguous, technical representation. Best practice: `eval(repr(obj)) == obj`.
- **`__str__`:** For **end-users**. Pretty, human-readable representation.
- **Fallback Rule:** If `__str__` is not implemented, Python automatically falls back to `__repr__`. If `__repr__` is not implemented, you get the default memory address `<... object at 0x...>`.

```python
# __str__ vs __repr__ Demonstration
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
        
    def __str__(self):
        return f"'{self.title}' by {self.author}"
        
    def __repr__(self):
        return f"Book(title={self.title!r}, author={self.author!r}, pages={self.pages})"

book = Book("Fluent Python", "Luciano Ramalho", 800)

print("str(book)  ->", str(book))   # User-facing
print("repr(book) ->", repr(book))  # Developer-facing (valid python code to recreate object)
```

### 2. Container Protocols: Emulating Built-in Collections
**Interview Question:** *How do you make your custom class support `len(obj)`, indexing `obj[i]`, and membership checks `'item' in obj`?*

- Implement `__len__`, `__getitem__`, `__setitem__`, and `__contains__`.

```python
# Custom SmartDeck Container Class
class CardDeck:
    ranks = [str(n) for n in range(2, 11)] + list('JQKA')
    suits = ['Spades', 'Diamonds', 'Clubs', 'Hearts']
    
    def __init__(self):
        self._cards = [(rank, suit) for suit in self.suits for rank in self.ranks]
        
    def __len__(self):
        return len(self._cards)
        
    def __getitem__(self, position):
        return self._cards[position]
        
    def __contains__(self, card):
        return card in self._cards

deck = CardDeck()
print(f"Total cards (via len): {len(deck)}")
print(f"First card (via deck[0]): {deck[0]}")
print(f"Top 3 cards (via slicing): {deck[:3]}")
print("Is ('A', 'Spades') in deck?", ('A', 'Spades') in deck)
```

### 3. Context Managers: `__enter__` and `__exit__`
**Interview Question:** *How does the `with` statement work in Python? What arguments does `__exit__` receive when an exception occurs?*

- `__enter__()`: Executed before the with-block, its return value is assigned to `as target`.
- `__exit__(self, exc_type, exc_val, exc_tb)`: Always executed upon leaving the block. Returning `True` suppresses exceptions; returning `False` allows them to propagate.

```python
# Custom Execution Timer Context Manager
import time

class ExecutionTimer:
    def __init__(self, task_name="Task"):
        self.task_name = task_name
        
    def __enter__(self):
        self.start_time = time.perf_counter()
        print(f"⏱️ Starting {self.task_name}...")
        return self
        
    def __exit__(self, exc_type, exc_val, exc_tb):
        elapsed = time.perf_counter() - self.start_time
        print(f"⏱️ Finished {self.task_name} in {elapsed:.6f} seconds.")
        if exc_type is not None:
            print(f"⚠️ Exception handled during {self.task_name}: {exc_val}")
        return True # Suppress exceptions safely

# Testing context manager
with ExecutionTimer("Matrix computation"):
    total = sum(x**2 for x in range(500_000))
```

### 4. Hashability and Set / Dict Keys: `__eq__` and `__hash__`
**Interview Question:** *Why can't you put an object in a `set` or use it as a `dict` key if you define `__eq__` without `__hash__`?*

- By default, Python objects are hashed by memory ID.
- Once you define `__eq__`, Python automatically sets `__hash__ = None` to prevent hashing mutable objects with changing equality.
- To make a custom class usable as a dictionary key or in a set, you must explicitly implement a consistent `__hash__()` based on immutable attributes.

```python
class UserAccount:
    def __init__(self, user_id, username):
        self.user_id = user_id
        self.username = username
        
    def __eq__(self, other):
        if isinstance(other, UserAccount):
            return self.user_id == other.user_id
        return False
        
    def __hash__(self):
        # Must hash the same immutable attributes used in __eq__
        return hash(self.user_id)

user1 = UserAccount(101, "Alice")
user2 = UserAccount(101, "Alice_Updated")

# Both have user_id 101 -> treated as equal in a set!
unique_users = {user1, user2}
print("Total unique accounts in set (deduplicated by user_id):", len(unique_users))
```
