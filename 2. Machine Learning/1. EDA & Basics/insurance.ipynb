```python
df_cleaned.head()
```

```python
from scipy.stats import pearsonr

# ----------------------------------
# Pearson Correlation Calculation
# ----------------------------------

# List of features to check against target
selected_features = [
    'age', 'bmi', 'children', 'is_female', 'is_smoker',
    'region_northwest', 'region_southeast', 'region_southwest',
    'bmi_category_Normal', 'bmi_category_Overweight', 'bmi_category_Obese'
]

correlations = {
    feature: pearsonr(df_cleaned[feature], df_cleaned['charges'])[0]
    for feature in selected_features
}
correlation_df = pd.DataFrame(list(correlations.items()), columns=['Feature', 'Pearson Correlation'])
correlation_df.sort_values(by='Pearson Correlation', ascending=False)
```

```python
cat_features = [
    'is_female', 'is_smoker',
    'region_northwest', 'region_southeast', 'region_southwest',
    'bmi_category_Normal', 'bmi_category_Overweight', 'bmi_category_Obese'
]
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05

df_cleaned['charges_bin'] = pd.qcut(df_cleaned['charges'], q=4, labels=False)
chi2_results = {}

for col in cat_features:
    contingency = pd.crosstab(df_cleaned[col], df_cleaned['charges_bin'])
    chi2_stat, p_val, _, _ = chi2_contingency(contingency)
    decision = 'Reject Null (Keep Feature)' if p_val < alpha else 'Accept Null (Drop Feature)'
    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }

chi2_df = pd.DataFrame(chi2_results).T
chi2_df = chi2_df.sort_values(by='p_value')
chi2_df
```

### 1. Target Distribution & Log-Transformation
- Regression models (OLS) assume residuals are normally distributed, which is often violated when the target is heavily right-skewed.
- Applying $\log(y)$ compresses extreme values and stabilizes variance (homoscedasticity).

```python
plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
sns.histplot(df['charges'], kde=True, color='crimson')
plt.title(f'Original Charges (Skewness: {df["charges"].skew():.2f})', fontweight='bold')

plt.subplot(1, 2, 2)
sns.histplot(np.log1p(df['charges']), kde=True, color='teal')
plt.title(f'Log-Transformed Charges (Skewness: {np.log1p(df["charges"]).skew():.2f})', fontweight='bold')
plt.tight_layout()
plt.show()
```

### 2. Interaction Features: The `bmi * smoker` Phenomenon
In real-world data, features often do not act independently. A smoker with high BMI faces exponentially higher medical risks than an obese non-smoker or a normal-weight smoker alone.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='bmi', y='charges', hue='smoker', data=df, palette={'yes': 'red', 'no': 'blue'}, alpha=0.7)
plt.title('Charges vs BMI stratified by Smoker Status', fontsize=14, fontweight='bold')
plt.axvline(x=30, color='gray', linestyle='--', label='Obesity Threshold (BMI=30)')
plt.legend()
plt.show()

# Create the engineered interaction term
final_df_interactive = final_df.copy()
final_df_interactive['bmi_smoker_interaction'] = final_df_interactive['bmi'] * final_df_interactive['is_smoker']
print("Interaction feature created! Correlation with charges:", final_df_interactive['bmi_smoker_interaction'].corr(final_df_interactive['charges']))
```

### 3. Multicollinearity Check with VIF
Evaluating whether any features exhibit dangerous linear dependency before feeding into multiple linear regression.

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
from sklearn.preprocessing import StandardScaler

X_features = final_df.drop('charges', axis=1)
X_scaled_vif = StandardScaler().fit_transform(X_features)

vif_table = pd.DataFrame({
    'Feature': X_features.columns,
    'VIF': [variance_inflation_factor(X_scaled_vif, i) for i in range(X_features.shape[1])]
}).sort_values(by='VIF', ascending=False)

print("=== VARIANCE INFLATION FACTOR (VIF) ===")
display(vif_table)
```

### 4. Pearson vs Spearman Correlation Matrix
- **Pearson**: Linear associations
- **Spearman**: Monotonic ranking associations (robust against non-linear curvatures)

```python
plt.figure(figsize=(14, 6))
plt.subplot(1, 2, 1)
sns.heatmap(final_df.corr(method='pearson'), annot=True, fmt='.2f', cmap='Blues', cbar=False)
plt.title('Pearson Correlation (Linear)', fontweight='bold')

plt.subplot(1, 2, 2)
sns.heatmap(final_df.corr(method='spearman'), annot=True, fmt='.2f', cmap='Greens', cbar=False)
plt.title('Spearman Rank Correlation (Monotonic)', fontweight='bold')
plt.tight_layout()
plt.show()
```

### 5. Regression EDA Interview Cheat Sheet

- **Why do we check target skewness in regression?**
  - While OLS does not strictly require the raw target $Y$ to be normal, having a heavily right-skewed target often induces heteroscedasticity and non-normal residuals. Log-transforming $Y$ produces well-behaved, homoscedastic errors.
- **How does an interaction term differ from polynomial features?**
  - Polynomial features ($X_1^2, X_1^3$) model non-linear responses of a single variable.
  - Interaction terms ($X_1 \times X_2$) model synergy or moderation where the effect of $X_1$ on $Y$ changes depending on the level of $X_2$.
- **What does a VIF < 5 indicate?**
  - It confirms that predictors share negligible linear collinearity, ensuring stable, reliable coefficients with low standard errors in Multiple Linear Regression.
